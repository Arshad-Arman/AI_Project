**Q2.write a program to display    a.number of rows and columns,   b.columns names   c.data types***


```python
#import Libraries
import pandas as pd
from sklearn.datasets import load_iris

# Load the Iris dataset
iris = load_iris(as_frame=True)

# Convert to a pandas DataFrame
df = iris.frame

# Display the number of rows and colimns
print("Number of Rows and columns:")
print(df.shape)

# Display the column names
print("\nColumn Names:")
print(df.columns)

# Display the data types
print("\nData Types:")
print(df.dtypes)
```

**Q3.write a program to count number of sample in iris datasets and digits datasets**

```python

from sklearn.datasets import load_iris, load_digits

# Load datasets
iris = load_iris()
digits = load_digits()

# Count the number of samples
print("Number of samples in Iris dataset:", len(iris.data))
print("Number of samples in Digits dataset:", len(digits.data))
```
